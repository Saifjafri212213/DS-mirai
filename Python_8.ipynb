```python
import pandas as pd
```

```python
df1 = pd.DataFrame({
    'id': [1,2,3,4],
    'name': ['Ali','Bhavesh','Chandu','Dhruv'],
    'age': [35,40,45,50]
})

df2 = pd.DataFrame({
    'id': [3,4,5,6],
    'name': ['Alok','Bhavan','Charlie','Dravid'],
    'age': [35,40,45,50]
})

# df3 = pd.merge(df1,df2,on='id',how='inner') #inner joint
# print(df3)

# df3.rename(columns={'name_x':'table_name1'},inplace=True)
df4 = pd.merge(df1,df2,on='id',how='left') #left joint
print(df4)

# df5 = pd.merge(df1,df2,on='id',how='right') #right joint
# print(df5)

df6 = pd.merge(df1,df2,how='cross') #cross joint
print(df6)
#
```
