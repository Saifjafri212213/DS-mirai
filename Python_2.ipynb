***List***:

```python
l = [1,2,34,'a','d', "hello",[2,3]]
print(l)

l.append([1,2.3])
print(l)

l.extend([1,2,3])
print(l)

l.insert(2 ,"insert")
print(l)

l.pop()
print(l)

l.pop(2)
print(l)

l.remove(34)
print(l)

print(l.index(2))

print(l.count(2))

```

```python
lst = [2,54,5,1,9,7,12,4]
lst.sort()
print(lst)

ls = sorted(lst)
print(ls)

#sum
print(sum(lst))

#min
print(min(lst))

#max
print(max(lst))
```
