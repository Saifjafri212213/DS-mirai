```python
stocks = np.random.randint(20,80,(5,5))
print(stocks)

demand = np.random.randint(20,80,(5,5))
print(demand)

remaining = stocks - demand
print(remaining)

c = np.where(remaining<0, 0, remaining)
print(c)

s = np.sum(remaining[remaining<0])
print(s*-1)

f = np.count_nonzero(remaining<0, axis=0)
print(f)

g = np.sort(remaining, axis=0)
print(g )


```
